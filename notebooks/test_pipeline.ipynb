```python
import asyncio

async def test_pipeline(ticker):
    orch = OrchestratorAgent()
    state = await orch.run(ticker)
    print()
    print('=== PIPELINE RESULT ===')
    print(f'Ticker:          {state.ticker}')
    print(f'Request ID:      {state.request_id}')
    print(f'Price rows:      {state.price_rows_available}')
    print(f'News articles:   {state.news_articles_count}')
    print(f'Has financials:  {state.has_recent_financials}')
    print(f'Quality score:   {state.data_quality_score:.3f}')
    print(f'Final weights:   {state.weights}')
    print(f'Direction:       {state.final_direction}')
    print(f'Data output:     {state.agent_outputs.get("data")}')
    print()
    print('--- Execution Log ---')
    for entry in state.execution_log:
        print(entry)
    return state

state = await test_pipeline('AMEN BANK')
```

```python
# Run pipeline for several stocks and compare quality scores + weights
# Order chosen to highlight Day 8 divergence: banks first, then non-banks
test_tickers = ['AMEN BANK', 'BIAT', 'BNA', 'SFBT', 'POULINA GP HOLDING']

results = []
for ticker in test_tickers:
    orch = OrchestratorAgent()
    s = await orch.run(ticker)
    results.append({
        'ticker': ticker,
        'price_rows': s.price_rows_available,
        'news': s.news_articles_count,
        'has_fin': s.has_recent_financials,
        'quality': s.data_quality_score,
        'w_tech': s.weights['technical'],
        'w_fund': s.weights['fundamental'],
        'w_sent': s.weights['sentiment'],
    })

import pandas as pd
df = pd.DataFrame(results)
print(df.to_string(index=False))
```
